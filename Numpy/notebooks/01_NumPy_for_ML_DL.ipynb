## 6. Slicing: selecting ranges

General form: `array[start:stop:step]`. The `stop` index is excluded.
- `a[:3]`: first three elements.
- `a[2:]`: from index 2 to the end.
- `a[::2]`: every second element.
- For 2D arrays, `array[row_slice, column_slice]`.

````python
a = np.arange(10, 21)
print("Original:", a)
print("Indices 2 through 5:", a[2:6])
print("First four:", a[:4])
print("From index 7:", a[7:])
print("Every second item:", a[::2])
print("Reversed:", a[::-1])
````

````python
grid = np.arange(1, 26).reshape(5, 5)
print("Grid:\n", grid)
print("\nTop-left 2×2:\n", grid[:2, :2])
print("\nAll rows, columns 1 to 3:\n", grid[:, 1:4])
print("\nRows 1 to 3, all columns:\n", grid[1:4, :])
print("\nEvery other row and column:\n", grid[::2, ::2])
````

### Slicing a 4D image batch
Suppose the shape is `(batch, height, width, channels)`.
- `images[0]`: one image.
- `images[:, :, :, 0]`: red-channel plane for every image.
- `images[:, 0:1, :, :]`: keep only the first row but preserve a height axis of length 1.

````python
print("First image:", images[0].shape)
print("Red channel for every image:", images[:, :, :, 0].shape)
print("First row of every image, axis preserved:", images[:, 0:1, :, :].shape)
````

## 7. Slice views vs copies — a common beginner trap

Basic slicing often returns a **view** into the original array. Editing the view can edit the original. Use `.copy()` when you need independent data.

Advanced indexing (for example, selecting with a list of integer indices or a Boolean mask) generally returns a copy.

````python
original = np.array([10, 20, 30, 40, 50])
view = original[1:4]
view[0] = 999
print("View:", view)
print("Original changed:", original)

independent = original[1:4].copy()
independent[0] = -100
print("Independent copy:", independent)
print("Original unchanged by copy edit:", original)
````

## 10. Reshaping arrays

`reshape()` changes the way the same sequence of elements is organized into axes. The total number of elements must remain constant.

For example, 12 elements can become `(3, 4)`, `(2, 2, 3)`, or `(1, 12)`, but not `(5, 3)`.

Official reference: [numpy.reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html)

````python
a = np.arange(12)
print("Original:", a, "shape =", a.shape)

as_3_by_4 = a.reshape(3, 4)
as_2_by_2_by_3 = a.reshape(2, 2, 3)
as_row = a.reshape(1, 12)

print("\n3×4:\n", as_3_by_4)
print("Shape:", as_3_by_4.shape)
print("\n2×2×3 shape:", as_2_by_2_by_3.shape)
print("Row shape:", as_row.shape)
````

### Using `-1` in reshape
One dimension may be inferred automatically using `-1`. NumPy calculates the missing size from the total element count and the other dimensions.

````python
a = np.arange(24)
print(a.reshape(2, -1).shape)       # 2 rows, infer columns
print(a.reshape(-1, 3, 4).shape)    # infer first dimension
````

### Common ML/DL reshaping examples
- Flatten a 2D image to a feature vector: `image.reshape(-1)`.
- Turn `N` examples with `F` features into a matrix: `(N, F)`.
- Add a batch dimension to one image: `image[np.newaxis, ...]`.
- Convert a vector to a column: `vector.reshape(-1, 1)`.

**Important:** reshape does not understand what your data means. It only reorganizes the element sequence. Verify that your intended axis meanings are preserved.

````python
one_sample = np.array([0.2, 0.4, 0.8])
batch_of_one = one_sample[np.newaxis, :]
column_vector = one_sample.reshape(-1, 1)

print("Sample:", one_sample.shape)
print("Batch of one:", batch_of_one.shape)
print("Column vector:", column_vector.shape)
````

## 14. Broadcasting

Broadcasting lets NumPy operate on compatible shapes without manually repeating values. Compare dimensions from right to left: dimensions are compatible when they are equal or one of them is 1. Missing leading dimensions behave as 1.

Official guide: [NumPy broadcasting](https://numpy.org/doc/stable/user/basics.broadcasting.html)

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
print("Add scalar:\n", data + 5)

column_offsets = np.array([[100], [200]])  # shape (2, 1)
print("Add one offset per row:\n", data + column_offsets)

feature_offsets = np.array([1, 2, 3])      # shape (3,)
print("Add one offset per column:\n", data + feature_offsets)
````

### Broadcasting shape reasoning
- `(2, 3)` plus `(3,)` → compatible, result `(2, 3)`.
- `(2, 3)` plus `(2, 1)` → compatible, result `(2, 3)`.
- `(2, 3)` plus `(2,)` → usually incompatible because the trailing dimensions are 3 and 2.

Before running an operation, write the shapes on paper and compare from the rightmost dimension.

````python
matrix = np.ones((2, 3))
vector3 = np.ones((3,))
vector2 = np.ones((2,))

print("Compatible:", (matrix + vector3).shape)
try:
    matrix + vector2
except ValueError as error:
    print("Incompatible shapes:", error)
````

## 15. Aggregations and the meaning of `axis`

Common reductions:
- `np.sum`, `np.mean`, `np.median`
- `np.min`, `np.max`
- `np.std`, `np.var`
- `np.argmax`, `np.argmin`

For a 2D array:
- `axis=0` reduces down the rows and returns one result per column.
- `axis=1` reduces across the columns and returns one result per row.
- no axis specified: reduce all elements.

````python
marks = np.array([
    [82, 91, 88],
    [75, 84, 90],
    [93, 89, 96],
    [68, 72, 80]
])

print("Mean of all marks:", marks.mean())
print("Mean per subject (column):", marks.mean(axis=0))
print("Mean per student (row):", marks.mean(axis=1))
print("Highest mark per subject:", marks.max(axis=0))
print("Best subject index per student:", marks.argmax(axis=1))
````

## 17. Missing values and data types

NumPy often uses `np.nan` to represent missing floating-point values. Ordinary `mean` propagates NaN; `np.nanmean` ignores NaNs. `np.nan` is not the same as Python `None`.

Integer arrays cannot directly store `np.nan` without conversion to a floating-point dtype.

````python
readings = np.array([20.5, 21.0, np.nan, 22.5, 23.0])
print("Ordinary mean:", np.mean(readings))
print("Mean ignoring NaN:", np.nanmean(readings))
print("Missing-value mask:", np.isnan(readings))
````

## 19. Linear algebra: dot products and matrix multiplication

For ML, feature matrices and weight vectors often interact through matrix multiplication.

If `X` has shape `(number_of_samples, number_of_features)` and `w` has shape `(number_of_features,)`, then `X @ w` has shape `(number_of_samples,)`.

Official reference: [NumPy linear algebra](https://numpy.org/doc/stable/reference/routines.linalg.html)

````python
X = np.array([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 4.0]
])  # 3 samples, 2 features

w = np.array([10.0, 100.0])  # one weight per feature
predictions_without_bias = X @ w

print("X shape:", X.shape)
print("w shape:", w.shape)
print("Predictions:", predictions_without_bias)
````

### Element-wise multiplication vs matrix multiplication
For same-shaped arrays, `A * B` multiplies corresponding elements. `A @ B` combines rows and columns according to linear algebra rules.

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])

print("Element-wise A * B:\n", A * B)
print("Matrix product A @ B:\n", A @ B)
````

## 20. Practical ML example 1 — feature scaling

### Problem
A dataset contains two features:
- hours studied (roughly 1–10)
- annual income (roughly thousands)

The scales are very different. Min-max scaling maps each feature into the range `[0, 1]` using:

\[
x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

For a feature matrix with rows = samples and columns = features, calculate minima and maxima **column-wise** with `axis=0`.

````python
X = np.array([
    [1.0, 30000.0],
    [3.0, 50000.0],
    [5.0, 70000.0],
    [7.0, 90000.0]
])

minimum = X.min(axis=0)
maximum = X.max(axis=0)
range_ = maximum - minimum

X_scaled = (X - minimum) / range_

print("Original X:\n", X)
print("Column minima:", minimum)
print("Column maxima:", maximum)
print("Scaled X:\n", X_scaled)
````

**Step-by-step interpretation**
1. `X.min(axis=0)` finds the minimum of each feature.
2. `X - minimum` broadcasts those minima across every sample row.
3. Division by `range_` scales each feature independently.
4. The output still has shape `(4, 2)`.

**Production note:** If a feature has zero range, division by zero occurs. In a real pipeline, handle constant features explicitly and fit scaling parameters on the training data only, then reuse them for validation/test data.

## 21. Practical ML example 2 — a simple linear prediction model

### Problem
Predict a house price from its size using a toy linear model:

\[
\hat{y} = Xw + b
\]

Here each row is one house and the single feature is size in square metres. The chosen weights are illustrative, not trained from real data.

````python
# Feature matrix: 3 houses, 1 feature (area in 100s of square metres)
X = np.array([[8.0], [10.0], [12.0]])
w = np.array([15.0])
b = 20.0

y_pred = X @ w + b

print("X shape:", X.shape)
print("w shape:", w.shape)
print("Predicted toy prices (arbitrary units):", y_pred)
````

### Calculate mean absolute error (MAE)
Given actual values and predictions, MAE is the mean absolute difference. It is a common regression metric.

````python
y_true = np.array([135.0, 175.0, 190.0])
mae = np.mean(np.abs(y_true - y_pred))
print("Predictions:", y_pred)
print("Actual:", y_true)
print("Absolute errors:", np.abs(y_true - y_pred))
print("MAE:", mae)
````

## 22. Practical ML example 3 — a tiny gradient-descent update

This is a teaching example showing why array operations matter in model training. We fit a one-feature linear model `prediction = X @ w + b` and update parameters using the gradient of mean squared error (MSE).

For \(m\) samples:
\[
MSE = \frac{1}{m}\sum_i (\hat y_i-y_i)^2
\]

For this model:
\[
\frac{\partial MSE}{\partial w} = \frac{2}{m}X^T(\hat y-y), \quad
\frac{\partial MSE}{\partial b} = \frac{2}{m}\sum_i(\hat y_i-y_i)
\]

The code performs a few updates. It is deliberately small enough to inspect.

````python
X = np.array([[1.0], [2.0], [3.0], [4.0]])
y = np.array([3.0, 5.0, 7.0, 9.0])  # roughly y = 2x + 1

w = np.array([0.0])
b = 0.0
learning_rate = 0.01
m = len(y)

for step in range(1000):
    y_hat = (X @ w).ravel() + b
    error = y_hat - y
    dw = (2 / m) * (X.T @ error)
    db = (2 / m) * np.sum(error)
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Learned weight:", w)
print("Learned bias:", b)
print("Predictions:", (X @ w).ravel() + b)
````

## 23. Practical DL example — image tensors and normalization

### Problem
A camera provides a batch of 32 RGB images of size 64×64. A common array shape is `(32, 64, 64, 3)` in NHWC layout.

For 8-bit image data, pixel values often range from 0 to 255. Dividing by 255 converts them to `[0, 1]`. This is a common preprocessing step, but the correct preprocessing must match the model's training setup.

````python
rng = np.random.default_rng(7)
image_batch_uint8 = rng.integers(0, 256, size=(32, 64, 64, 3), dtype=np.uint8)

print("Batch shape:", image_batch_uint8.shape)
print("Data type:", image_batch_uint8.dtype)
print("Minimum and maximum:", image_batch_uint8.min(), image_batch_uint8.max())

image_batch_float = image_batch_uint8.astype(np.float32) / 255.0
print("Normalized dtype:", image_batch_float.dtype)
print("Normalized min/max:", image_batch_float.min(), image_batch_float.max())
````

### Select a mini-batch and one channel

````python
mini_batch = image_batch_float[:8]
green_channel = image_batch_float[:, :, :, 1]

print("Mini-batch shape:", mini_batch.shape)
print("Green-channel batch shape:", green_channel.shape)
````

## 25. Common errors and debugging checklist

1. **Index out of bounds:** an index must be valid for that axis.
2. **Wrong reshape size:** total element count must match.
3. **Broadcasting error:** compare shapes from the rightmost axis.
4. **Unexpected mutation:** basic slices may be views; use `.copy()` if needed.
5. **Wrong `axis`:** explicitly state what each axis represents before reducing.
6. **Integer division assumptions:** `/` generally produces floating-point results.
7. **Confusing `*` and `@`:** element-wise multiplication versus matrix multiplication.
8. **Wrong image layout:** confirm NHWC vs NCHW.
9. **Integer overflow:** fixed-width integer dtypes have finite ranges.
10. **NaN propagation:** use `np.isnan`, `np.nanmean`, etc. when appropriate.

````python
# Debugging pattern: inspect the array before operating on it
sample = np.arange(24).reshape(2, 3, 4)
print("shape:", sample.shape)
print("ndim:", sample.ndim)
print("size:", sample.size)
print("dtype:", sample.dtype)
````
