# Pandas DataFrames: What They Are and How to Create Them

A **DataFrame** is a two-dimensional, labeled table in pandas. It has rows and columns, much like a spreadsheet or a database table. Each column is a pandas **Series**, and the row labels form the DataFrame's **index**.

In this lesson, you will compare a Series with a DataFrame, create DataFrames from common Python data structures, and inspect their rows, columns, and shape.

## Series and DataFrame: how are they different?

A **Series** holds one labeled sequence of values (one column). A **DataFrame** brings one or more Series together as columns in a single table. In the diagrams below, the country names are the shared row index.

The first image shows one DataFrame with `Year`, `Pop`, and `GDP` columns:

![A DataFrame containing Year, Pop, and GDP columns for each country](Images/Pandas_series_Same_Index.png)

The second image separates those columns into three Series. Each Series uses the same country index, so pandas can align them as columns in a DataFrame:

![Three Series with the same country index](Images/Same_index_df.png)

Think of a DataFrame as a collection of aligned Series: each row label identifies a record, and each column describes one variable.

## 1. Import pandas

The conventional alias for pandas is `pd`. We use `pd.DataFrame(...)` to create a DataFrame.

## 2. Create a DataFrame from a dictionary of lists

A common starting point is a dictionary where each **key** becomes a column name and each **list** provides that column's values. All lists must have the same length so each row has a value in every column.

```python
data = {
    "Country": ["USA", "Canada", "Mexico"],
    "Year": [1776, 1867, 1821],
    "Population (millions)": [328, 38, 126],
    "GDP (trillions)": [20.5, 1.7, 1.22],
}

countries_df = pd.DataFrame(data)
countries_df
```

The dictionary keys became the column labels. Since we did not supply an index, pandas created the default integer index `0, 1, 2`. The index labels rows; they are not an extra data column.

## 3. Create a DataFrame with a meaningful index

Pass an `index` to `pd.DataFrame` when the row labels should be meaningful. Here, country names identify rows, so we do not need to repeat them as a regular column.

```python
country_data = {
    "Year": [1776, 1867, 1821],
    "Pop": [328, 38, 126],
    "GDP": [20.5, 1.7, 1.22],
}

country_df = pd.DataFrame(
    country_data,
    index=["USA", "CANADA", "MEXICO"],
)
country_df
```

## 4. Create a DataFrame from a list of dictionaries

A list of dictionaries is useful when your data is already organized as **records**. Each dictionary describes one row; its keys identify columns. pandas matches values to columns by key.

```python
records = [
    {"Product": "Notebook", "Price": 3.50, "In stock": True},
    {"Product": "Pen", "Price": 1.25, "In stock": True},
    {"Product": "Folder", "Price": 2.00, "In stock": False},
]

products_df = pd.DataFrame(records)
products_df
```

## 5. Create a DataFrame from Series

You can also combine Series as columns. pandas aligns their values by index label, which is why the Series in the illustration above need the same index when their values should appear on the same rows.

```python
year = pd.Series([1776, 1867, 1821], index=["USA", "CANADA", "MEXICO"], name="Year")
population = pd.Series([328, 38, 126], index=["USA", "CANADA", "MEXICO"], name="Pop")

series_df = pd.DataFrame({"Year": year, "Pop": population})
series_df
```

## 6. Inspect a DataFrame

These properties and methods help you understand a DataFrame's structure:

- `shape` gives the number of rows and columns.
- `columns` lists the column labels, and `index` gives the row labels.
- `dtypes` shows the data type of each column.
- `head()` previews the first rows.

```python
print("Shape (rows, columns):", country_df.shape)
print("Columns:", list(country_df.columns))
print("Index:", list(country_df.index))
print("\nColumn data types:\n", country_df.dtypes)
print("\nFirst rows:\n", country_df.head())
```

## Recap and practice

- A DataFrame is a labeled, two-dimensional table; its columns are Series and its rows are identified by an index.
- `pd.DataFrame(...)` can create a DataFrame from a dictionary of lists, a list of row dictionaries, or Series.
- Dictionary keys become column names. Provide `index=` when you want specific row labels.
- Values in Series are aligned by index labels when they are combined into a DataFrame.
