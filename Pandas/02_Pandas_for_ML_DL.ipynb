# Pandas Series: A Step-by-Step Guide

A **Series** is a one-dimensional pandas object that stores values together with an index. Think of it as a column of data with labels. We will build a Series of countries and their years of independence, then retrieve values using labels and positions.

By the end, you will be able to:
- create a Series from a list;
- replace the default numeric index with country-name labels;
- name both the Series and its index;
- retrieve one or more values using common indexing styles; and
- use useful Series attributes and methods.

> Historical note: the years below use commonly recognized independence dates for these examples. The meaning and date of independence can be historically nuanced.

## 1. Import pandas

The conventional alias for pandas is `pd`. This notebook uses the `CUDA_ENV` Python environment, where pandas is installed.

## 2. Create a Series with the default index

Pass values to `pd.Series(...)`. If no index is supplied, pandas creates a `RangeIndex` starting at zero. Here each position happens to correspond to a country in the separate `countries` list. Use `.iloc` to retrieve by position; positions are zero-based.

```python
countries = ["India", "Ghana", "Kenya", "Singapore", "Bangladesh"]
years = [1947, 1957, 1963, 1965, 1971]

years_by_position = pd.Series(years)
print(years_by_position)
print("Index type:", type(years_by_position.index).__name__)

# The first country is at position 0; the fifth is at position 4.
print("Year at position 0:", years_by_position.iloc[0])
print("Year at position 4:", years_by_position.iloc[4])
```

## 3. Add meaningful labels to the index

A labeled index lets you look up a year by country instead of remembering its numeric position. Give the Series the name **`Year of independence`** and the index the name **`Country`**. These names describe different things: `.name` labels the Series, while `index.name` labels its index.

```python
independence_year = pd.Series(
    years,
    index=pd.Index(countries, name="Country"),
    name="Year of independence",
)

independence_year
```

## 4. Retrieve values using different indexing methods

- **`.loc[label]`** retrieves by index label, such as a country name.
- **`.iloc[position]`** retrieves by zero-based position.
- **`series[label]`** also retrieves a value by label here; explicit `.loc` and `.iloc` are clearer and avoid ambiguity.
- A list of labels selects several rows. Label slices with `.loc` include the end label; positional slices with `.iloc` exclude the end position.
- A boolean condition selects rows whose values meet a rule.

```python
# One value by label or by position
print("By label (.loc):", independence_year.loc["Kenya"])
print("By position (.iloc):", independence_year.iloc[2])
print("Bracket label lookup:", independence_year["Kenya"])

# Several labels, a label slice (inclusive), and a position slice (end-exclusive)
print("\nSelected countries:\n", independence_year.loc[["India", "Singapore"]])
print("\nLabel slice:\n", independence_year.loc["Ghana":"Singapore"])
print("\nPosition slice:\n", independence_year.iloc[1:4])

# Boolean filtering
print("\nIndependence year >= 1960:\n", independence_year.loc[independence_year >= 1960])
```

## 5. Useful Series methods

Methods calculate or return useful views of the values. For example, `head()` previews the beginning, `sort_values()` orders by year, `describe()` summarizes numeric data, and `idxmax()` returns the label for the largest value. Methods such as `sort_values()` return a result; they do not change the original Series unless you assign the result.

```python
print("First three rows:\n", independence_year.head(3))
print("\nLast two rows:\n", independence_year.tail(2))
print("\nSorted by year:\n", independence_year.sort_values())
print("\nSorted by country:\n", independence_year.sort_index())
print("\nNumeric summary:\n", independence_year.describe())
```

```python
# Scalar methods and useful conversions
print("Latest year in this sample:", independence_year.max())
print("Country with the latest year:", independence_year.idxmax())
print("Average year in this sample:", independence_year.mean())
print("Is Ghana in the index?", "Ghana" in independence_year.index)
print("\nCountries matching selected years:\n", independence_year.loc[independence_year.isin([1947, 1957])])
print("As a dictionary:", independence_year.to_dict())

# Attributes describe the Series without calling a method.
print("\nSeries name:", independence_year.name)
print("Index name:", independence_year.index.name)
print("Data type:", independence_year.dtype)
print("Number of values:", independence_year.size)
```
