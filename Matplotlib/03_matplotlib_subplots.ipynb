## Pandas Plotting and Matplotlib Subplots

### 1. What is Pandas Plotting?

Pandas provides built-in plotting methods that use Matplotlib internally. These methods allow us to visualize data directly from a Pandas `Series` or `DataFrame` without manually extracting every column.

**Typical syntax:**

```python
data.plot(kind="line", x=None, y=None, figsize=(8, 5),
          title="My Plot", color="blue", grid=True)
```

Common arguments include:

- `kind`: Specifies the plot type, such as `"line"`, `"bar"`, `"scatter"`, `"hist"`, `"box"`, `"area"`, or `"pie"`.
- `x`: Column used for the horizontal axis.
- `y`: Column or list of columns to visualize.
- `figsize`: Figure dimensions in inches, for example `(8, 5)`.
- `title`: Title displayed above the chart.
- `color`: Specifies the plot color.
- `grid`: Enables or disables grid lines.
- `legend`: Controls whether a legend is displayed.

Example:

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [100, 150, 130, 200],
    "Expenses": [60, 80, 75, 110]
})

df.plot(x="Month", y=["Sales", "Expenses"],
        kind="line", marker="o", grid=True)

plt.show()
```

### 2. What Are Matplotlib Subplots?

Subplots allow us to display multiple charts inside a single figure. They are useful for comparing different variables or visualization techniques side by side.

**Typical syntax:**

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
```

- `nrows`: Number of rows of plots.
- `ncols`: Number of columns of plots.
- `figsize`: Overall figure size.
- `fig`: The complete figure containing all charts.
- `axes`: Array of individual Axes objects on which plots are drawn.

Example:

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

df.plot(x="Month", y="Sales", ax=axes[0], kind="bar")
df.plot(x="Month", y="Expenses", ax=axes[1], kind="line", marker="o")

axes[0].set_title("Monthly Sales")
axes[1].set_title("Monthly Expenses")

plt.tight_layout()
plt.show()
```

**Key concept:** Pandas `plot()` creates a chart, while `plt.subplots()` creates the layout. The `ax` argument tells Pandas which subplot should receive the chart.
