```python
np.random.seed(101)
x = np.random.randint(1,12,5)
x
```

```python
y = 2*x 
y
```

```python
y_squared = x**2  
y_squared
```

```python
plt.plot(x,y)
```

```python
fig = plt.figure()

axes_1 = fig.add_axes([0,0,1,1])
axes_1.plot(x,y,label="X vs Y -Linear relationship",color= "red")

axes_2 = fig.add_axes([-0.25,0.35,0.25,0.25])
axes_2.plot(x,y_squared,label="x vs square_x - 2nd degree polynomial")

plt.legend()

plt.show()
```

```python
fig = plt.figure()

axes_1 = fig.add_axes([0,0,1,1])
axes_1.plot(x,y)


axes_2 = fig.add_axes([0,0,0.25,0.25])
axes_2.plot(x,y_squared)



```

```python
fig = plt.figure(figsize=(12,8),dpi=720)

axes_1 = fig.add_axes([0.25,0,1,1])
axes_1.plot(x,y)


axes_2 = fig.add_axes([0,0,0.25,0.25])
axes_2.plot(x,y_squared)

fig.savefig("new_axes_objects.png",bbox_inches='tight')

```

```python
fig = plt.figure(figsize=(12,8),dpi=720)

axes_1 = fig.add_axes([0,0,1,1])
axes_1.plot(x,y,label = "X vs Y")


#axes_2 = fig.add_axes([0,0,0.25,0.25])
axes_1.plot(x,y_squared, label = "X vs y_squared")

#fig.savefig("new_axes_objects.png",bbox_inches='tight')
fig.show()
plt.legend(loc =(-1,1))
```

```python
fig = plt.figure(figsize=(10,6),dpi=720)

axes_1 = fig.add_axes([0,0,1,1])
axes_1.plot(x,y,label="label infor 1")
axes_1.set_xlabel("X-axis")
axes_1.set_ylabel("y-axis")

axes_2 = fig.add_axes([0,0.25,0.50,0.75])

axes_2.plot(x,x**3, label="x vs cube of x")
plt.tight_layout()

plt.show()
```

```python
fig = plt.figure(figsize=(10,6),dpi=720)

axes_1 = fig.add_axes([0,0,1,1])
axes_1.plot(x,y,label="label infor 1")
axes_1.set_xlabel("X-axis")
axes_1.set_ylabel("y-axis")

axes_2 = fig.add_axes([0,0.25,0.50,0.75])

axes_2.plot(x,x**3, label="x vs cube of x")



axes_3 = fig.add_axes([0,0.25,0.50,0.55])
axes_3.plot(x,x**3 - x**2, label="complex curve")
axes_3.set_xlim(0,100)

plt.tight_layout()

plt.show()
```
